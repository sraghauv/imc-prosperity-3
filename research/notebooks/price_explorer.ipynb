```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

```python
## Loading the Data
csv_files = [
    '../../data/round1/prices_round_1_day_-2.csv',
    '../../data/round1/prices_round_1_day_-1.csv',
    '../../data/round1/prices_round_1_day_0.csv'
]
dataframes = [pd.read_csv(file, sep=';') for file in csv_files]
```

```python
# Concatenate all DataFrames into a single DataFrame
combined_df = pd.concat(dataframes, ignore_index=True)

# Convert 'day' and 'timestamp' columns to numeric in case they're read as strings
combined_df['day'] = pd.to_numeric(combined_df['day'], errors='coerce')
combined_df['timestamp'] = pd.to_numeric(combined_df['timestamp'], errors='coerce')
```

```python
time_offset = 1000000

# Use the minimum day value as the baseline so that the earliest day starts at zero offset
min_day = combined_df['day'].min()

# Create a new 'global_time' column that adjusts the timestamp based on the day value
combined_df['global_time'] = (combined_df['day'] - min_day) * time_offset + combined_df['timestamp']
```

Changing Database to only Show one Stock


```python

stock = 'RAINFOREST_RESIN'
rainforest_df = combined_df[combined_df['product'] == stock]
```

```python
if rainforest_df.empty:
    print("No data found for product 'RAINFOREST_RESIN'.")
else:
    # Set up a larger, higher resolution figure for clarity
    plt.figure(figsize=(16, 8), dpi=100)
    
    # Plot the mid_price using a clean line with no markers
    plt.plot(rainforest_df['global_time'], rainforest_df['mid_price'], linestyle='-', linewidth=1)
    
    # Label the axes and set the title
    plt.xlabel('Global Time (Adjusted)')
    plt.ylabel('Mid Price')
    plt.title('Mid Price Over Adjusted Global Time for ' + stock)
    
    # Rotate x-tick labels to avoid overlap
    plt.xticks(rotation=45)
    
    # Limit the number of x-axis ticks to keep the graph uncluttered
    ax = plt.gca()
    ax.xaxis.set_major_locator(ticker.MaxNLocator(10))
    
    # Enable grid lines for better readability
    plt.grid(True)
    plt.tight_layout()
    
    # Optionally, add interactive hover tooltips using mplcursors
    try:
        import mplcursors
        mplcursors.cursor(hover=True)
    except ImportError:
        print("mplcursors not installed; interactive tooltips are disabled.")
    
    plt.show()
```
